```python
# Importing required module
import pandas as pd
from tabulate import tabulate
import pprint
import json
from sqlalchemy import create_engine, text

print("Required module are successfully imported")
```

### **Reading csv file using python module pandas**

```python
dfc = pd.read_csv("/content/drive/MyDrive/raw_dataset/multiple_file_type/cust_info.csv")
```

### **Reading parquet file using python module pandas**

```python
dfp = pd.read_parquet("/content/drive/MyDrive/raw_dataset/multiple_file_type/sales_details.parquet")
print(tabulate(dfp.head(10), headers='keys', tablefmt='simple_grid'))
```

### **Reading orc file using python module pandas**

```python
dfo = pd.read_orc("/content/drive/MyDrive/raw_dataset/multiple_file_type/cust_info.orc")
print(tabulate(dfo.head(10), headers='keys', tablefmt='simple_grid'))
```

### **Reading json line file using python module pandas**

```python
dfj = pd.read_json("/content/drive/MyDrive/raw_dataset/multiple_file_type/taxi.jl", lines=True)
print(tabulate(dfj.head(10), headers='keys', tablefmt='simple_grid'))
```
