```python
# reading data using pandas
df = pd.read_csv("/content/drive/MyDrive/raw_dataset/retail_dataset/raw_customers.csv")
```

```python
# Data Decomposition
customer_identity = ['customer_id', 'title', 'first_name', 'last_name', 'full_name', 'gender', 'is_active']
customer_address = ['customer_id', 'address', 'city', 'state_abbr', 'state_full', 'zip_code', 'country', 'region']
customer_content = ['customer_id', 'email', 'phone']
customer_business_info = ['customer_id','customer_segment', 'loyalty_points','preferred_channel', 'annual_income_usd', 'company' ]
customer_dates = ['customer_id','date_of_birth', 'age', 'account_created_date']
```

```python
# defining the data frame for eatch decomposition 
customer_identity_df = df[customer_identity]
customer_address_df = df[customer_address]
customer_content_df = df[customer_content]
customer_business_info_df = df[customer_business_info]
customer_dates_df = df[customer_dates]
```

```python
# checking the issues in customer identity dataframe 
print(
    tabulate(
        customer_identity_df.sample(frac=0.02), 
        headers='keys', 
        tablefmt='simple_grid'
    )
)
```

```python
# checking the issues in customer address dataframe 
print(
    tabulate(customer_address_df.sample(frac=0.02),
            headers='keys', 
            tablefmt='simple_grid'
        )
)
```

```python
# checking the issues in customer content dataframe 
print(
    tabulate(
        customer_content_df.sample(frac=0.02), 
        headers='keys', 
        tablefmt='simple_grid'
    )
)
```

```python
# checking the issues in customer business information dataframe 
print(
    tabulate(
        customer_business_info_df.sample(frac=0.02), 
        headers='keys', 
        tablefmt='simple_grid'
    )
)
```

```python
# checking the issues in customer date's dataframe 
print(
    tabulate(
        customer_dates_df.sample(frac=0.02), 
        headers='keys', 
        tablefmt='simple_grid'
    )
)
```

```python
# checking the all numerici data type in the dataset 
print(
    tabulate(
        df.select_dtypes(include='number').sample(frac=0.02), 
        headers='keys', 
        tablefmt='simple_grid'
    )
)
```

```python
print(
    tabulate(
        customer_business_info_df.select_dtypes(include='object').sample(frac=0.02), 
        headers='keys', 
        tablefmt='simple_grid'
    )
)
```

### **Row Selection using pandas** 

```python
# Slicing by the positin 
print(
    tabulate(
        customer_identity_df[2:8], 
        headers='keys', 
        tablefmt='simple_grid'
    )
)
```

```python
# Boolean mask 
print(
    tabulate(
        customer_dates_df[customer_dates_df['age'] > 50].sample(frac=0.05),
        headers='keys',
        tablefmt='simple_grid'
    )
)
```

### **.loc — label-based (inclusive on both ends)**

```python
# selecting row with 0 inclusive lebel 
df.loc[0]
```

```python
# rows 0-5, cols name to score
print(
    tabulate(
        customer_dates_df.loc[0:5, 'customer_id' : 'date_of_birth'],
        headers='keys',
        tablefmt='simple_grid'
    )
)
```

```python
# specific rows & cols
print(
    tabulate(
        customer_identity_df.loc[[1, 3, 5, 9 , 33], ['customer_id', 'first_name']],
        headers='keys',
        tablefmt='simple_grid'
    )
)
```

### **.iloc — integer position-based (exclusive end)**

```python
# Returning the last row from the table 
print(
    tabulate(
        customer_content_df.iloc[[-1]],
        headers='keys',
        tablefmt='simple_grid'
    )
)
```

```python
# Returning the last column from the table 
customer_content_df.iloc[: , -1].head(10)
```

### **.xs — cross-section (great for MultiIndex)**

```python
import pandas as pd

df = pd.DataFrame({
    "customer_segment": ["bronze", "bronze", "silver", "gold"],
    "city": ["Patna", "Delhi", "Patna", "Mumbai"],
    "sales": [100, 200, 300, 400]
})

df = df.set_index(["customer_segment", "city"])

df.xs('bronze', level='customer_segment')
```
