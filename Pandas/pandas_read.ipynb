```python
import pandas as pd
import xml.etree.ElementTree as ET

file_path = "/content/drive/MyDrive/multiple_file_type/customers.xml"

# Parse XML
tree = ET.parse(file_path)
root = tree.getroot()

# Define namespace
ns = {'spreadsheet': 'http://schemas.openxmlformats.org/spreadsheetml/2006/main'}

# Find all rows
rows = root.findall('.//spreadsheet:row', ns)

data = []
headers = []

# Process each row
for i, row in enumerate(rows):
    row_data = {}
    
    # Find all cells in this row
    cells = row.findall('spreadsheet:c', ns)
    
    for cell in cells:
        # Get cell reference (A1, B1, etc.)
        cell_ref = cell.get('r')
        
        # Get cell value - it's inside <is><t> for inlineStr type
        inline_str = cell.find('spreadsheet:is', ns)
        if inline_str is not None:
            t = inline_str.find('spreadsheet:t', ns)
            if t is not None and t.text:
                value = t.text
            else:
                value = ''
        else:
            # Check for other cell types
            v = cell.find('spreadsheet:v', ns)
            if v is not None and v.text:
                value = v.text
            else:
                value = ''
        
        row_data[cell_ref] = value
    
    if i == 0:
        # First row contains headers
        headers = [row_data.get(f'{chr(65+j)}1', '') for j in range(len(row_data))]
    else:
        # Data rows - convert to list of values
        data.append([row_data.get(f'{chr(65+j)}{i+1}', '') for j in range(len(headers))])

# Create DataFrame
dfxml = pd.DataFrame(data, columns=headers)
new_table = dfxml[['customer_id', 'customer_number', 'first_name', 'last_name', 'country', 'marital_status', 'gender']]
print(tabulate(new_table.head(10), headers='keys', tablefmt='simple_grid'))
```

```python
import pandas as pd
from fastavro import writer, parse_schema

# Read CSV
df = pd.read_csv("/content/drive/MyDrive/multiple_file_type/oil_reserve_by_country.csv")

# Create Avro schema
schema = {
    "type": "record",
    "name": "OilReserve",
    "fields": [
        {"name": col, "type": ["null", "string"]}
        for col in df.columns
    ]
}

# Convert all values to string and handle nulls
records = (
    df.astype(str)
      .replace("nan", None)
      .to_dict(orient="records")
)

# Write Avro
with open("/content/oil_reserve_by_country.avro", "wb") as out:
    writer(out, parse_schema(schema), records)

print("Avro file created successfully!")
```
