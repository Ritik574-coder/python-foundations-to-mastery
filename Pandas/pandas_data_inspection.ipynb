```python
import pandas as pd 
from tabulate import tabulate 
```

```python
df = pd.read_csv("/content/drive/MyDrive/raw_dataset/retail_dataset/raw_customers.csv")
```

```python
pd.set_option('display.max_colwidth', None)
```

```python
customer_identity = ['customer_id', 'title', 'first_name', 'last_name', 'full_name', 'gender', 'is_active']
customer_address = ['customer_id', 'address', 'city', 'state', 'state_abbr', 'state_full', 'zip_code', 'country', 'region']
customer_content = ['customer_id', 'email', 'phone']
customer_business_info = ['customer_id','customer_segment', 'loyalty_points','preferred_channel', 'annual_income_usd', 'company' ]
customer_dates = ['customer_id','date_of_birth', 'age', 'account_created_date']
```

```python
customer_identity_df = df[customer_identity]
customer_address_df = df[customer_address]
customer_content_df = df[customer_content]
customer_business_info_df = df[customer_business_info]
customer_dates_df = df[customer_dates]
```

```python
# the summery of the datase
summery = pd.DataFrame({
    "data_type" : df.dtypes,
    "null_count": df.isnull().sum(),
    "null_percentage" : df.isnull().sum()/len(df),
    "not_null_count" : df.notnull().sum(),
    "unique_count" : df.nunique(),
    "duplicated_count" : df.apply(lambda col: col.duplicated().sum()),
    "null_like_values": df.apply(
    lambda col: col.astype(str).str.lower().isin(['null','none','n/a','na','']).sum()
    ),
    "memory_usage_mb": (df.memory_usage(deep=True)/1024**2).round(6)
})
print(tabulate(summery, headers='keys', tablefmt='simple_grid'))
```
