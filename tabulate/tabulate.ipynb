## **Using tabulate formats for improving output of pandas data frame** 

```python
# checking the all format of tabulate 
import tabulate
import pprint

format = tabulate.tabulate_formats
pprint.pp(format)
```

```python
print(tabulate.tabulate(df.head(10), headers='keys', tablefmt='plain'))
```
